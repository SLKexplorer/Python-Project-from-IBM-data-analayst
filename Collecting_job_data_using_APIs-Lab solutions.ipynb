```python
import pandas as pd
import json
```

```python
api_url="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"
df = pd.read_json(api_url)
```

```python
df.head()
```

```python
df.dtypes
```

```python
def get_number_of_jobs_T(technology):
    tech_df = df[df["Key Skills"].str.contains(technology)]
    return technology, len(tech_df.index)
```

```python
get_number_of_jobs_T("Python")
```

```python
def get_number_of_jobs_L(location):

    loc_df = df[df["Location"].str.contains(location)]
    return location, len(loc_df.index)
```

```python
get_number_of_jobs_L("Los Angeles")
```

```python
locs = [
    "Los Angeles",
    "New York",
    "San Francisco",
    "Washington DC",
    "Seattle",
    "Austin",
    "Detroit",
]
```

```python
loc_tuples = []
for loc in locs:
    loc_tuples.append(get_number_of_jobs_L(loc))

loc_df = pd.DataFrame(loc_tuples, columns=["Location", "Number of Jobs"])
loc_df.set_index("Location", inplace=True)
```

```python
locations = ['Los Angeles','New York','San Francisco','Washington DC','Seattle','Austin','Detroit']
```

```python
from openpyxl import Workbook
```

```python
wb1 = Workbook()
ws1 = wb1.active
```

```python
ws1.append(['Location','Number of Jobs'])

for i in range(len(locations)):
    ws1.append(get_number_of_jobs_L(locations[i]))
```

```python
wb1.save('2.a-job-postings (Collected from API).xlsx')
wb1.close()
```

```python
wb2 = Workbook()
ws2 = wb2.active
ws2.title = "Languages"

languages = ['C', 'C#', 'C++', 'Java', 'JavaScript', 'Python', 'Scala',
             'Oracle', 'SQL Server', 'MySQL Server', 'PostgreSQL', 'MongoDB']
def get_number_of_jobs_T(technology):
    tech_df = df[df["Key Skills"].str.contains(technology, case=False, na=False, regex=False)]
    return technology, len(tech_df.index)

ws2.append(['Languages', 'Number of Jobs'])

for lang in languages:
    ws2.append(get_number_of_jobs_T(lang))

wb2.save('2.a-job-postings-languages (Collected from API).xlsx')
```
